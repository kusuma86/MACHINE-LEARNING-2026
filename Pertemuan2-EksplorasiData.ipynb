# 15. Filtering Data

```python
# Misalnya mencari data Fault
df[df["condition"] == "Fault"]
```

```python
# Mencari temperatur di atas 50°C
df[df["temperature_c"] > 50]
```

```python
# berapa temperatur di atas 50C?
df[df["temperature_c"] > 50].shape[0]
```

### AND → &
### OR  → |

```python
# menghitung jumlah Fault yang temperaturnya di atas 50°C
df[
    (df["condition"] == "Fault") &
    (df["temperature_c"] > 50)
].shape[0]
```

```python
# menggunakan OR
df[
    (df["condition"] == "Warning") |
    (df["condition"] == "Fault")
]
```

# 25. Line Plot Berdasarkan Waktu

```python
plt.plot(df["timestamp"],df["temperature_c"])

plt.xlabel("Time")
plt.ylabel("Temperature (C)")
plt.title("Temperature Over Time")

plt.xticks(rotation=45)
plt.tight_layout()

plt.show()
```

# 26. Membandingkan Dua Sensor terhadap Waktu

```python
plt.plot(
    df["timestamp"],
    df["temperature_c"],
    label="Temperature"
)

plt.plot(
    df["timestamp"],
    df["current_a"],
    label="Current"
)

plt.xlabel("Time")
plt.title("Sensor Trends")
plt.legend()

plt.xticks(rotation=45)
plt.tight_layout()

plt.show()
```

# 27. Correlation Matrix sebagai Gambar

```python
corr = df.select_dtypes(include="number").corr()
```

```python
plt.figure(figsize=(10, 8))

plt.imshow(
    corr,
    aspect="auto",
    vmin=-1,
    vmax=1
)

plt.colorbar(label="Correlation")

plt.xticks(
    range(len(corr.columns)),
    corr.columns,
    rotation=90
)

plt.yticks(
    range(len(corr.columns)),
    corr.columns
)

plt.title("Correlation Matrix")
plt.tight_layout()

plt.show()
```
